```python
import numpy as np
import networkx as nx
import propnetscore.node_selection as ns
```

```python
G = nx.karate_club_graph()
```

```python
nx.draw(G, pos = nx.spring_layout(G), with_labels=True)
```

```python
G.nodes()
```

```python
G.edges()
```

```python
A = nx.to_numpy_array(G, weight=None)
```

```python
A
```

```python
import propnetscore.node_selection as ns

true_node = 4
prediction = np.zeros(len(G))
prediction[4] = 0.25
prediction[0] = 0.25
prediction[6] = 0.25
prediction[11] = 0.25

task1 = ns.NodeSelectionTask(A, prediction, true_node)
task1.brier_score()
task1.resistance_score()
task1.logarithmic_score()
task1.diffusion_score()
```
