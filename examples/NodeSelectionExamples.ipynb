```python
G = nx.karate_club_graph()
```

```python
adjacency_matrix = nx.to_numpy_array(G, weight=None)
```

```python
import propnetscore.node_selection as ns

# Ground truth
true_node = 4

# Prediction
prediction = np.zeros(len(G))
prediction[4] = 0.25
prediction[0] = 0.35
prediction[6] = 0.25
prediction[11] = 0.15

assert np.allclose(sum(prediction), 1)

task1 = ns.NodeSelectionTask(adjacency_matrix, prediction, true_node)

# Proper scoring rules
print("Brier Score: ", task1.brier_score())
print("Logarithmic Score: ", task1.logarithmic_score())
print("Resistance Score: ", task1.resistance_score())
print("Diffusion Score: ", task1.diffusion_score(1))
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Transform probabilities for better contrast
transformed_probs = task1.node_probs**(1/1)
transformed_probs = task1.node_probs/task1.node_probs.max()

# Normalize probabilities to [0,1] for the colormap
cmap = cm.Blues
colors = [cmap(p) for p in transformed_probs]

# Draw network with prediction and ground truth
pos = nx.spring_layout(G)
nx.draw_networkx_nodes(G, pos, node_color=colors, edgecolors='black', linewidths=1, node_size=500)
nx.draw_networkx_nodes(G, pos, nodelist=[task1.true_node], node_color=[colors[task1.true_node]], edgecolors='red', linewidths=4, node_size=500)
nx.draw_networkx_edges(G, pos)
nx.draw_networkx_labels(G, pos)

plt.axis('off')
plt.show()
```

```python
from propnetscore.plotting import plot_task

fig, ax = plt.subplots(figsize=(6, 6))

plot_task(
    ax, G, node_probs=prediction, true_node=true_node, 
    labels=False, colorbar=True, contrast=1, prob_color="Greens", true_color="Black", 
    scores=[
        "brier", 
        "laplace", 
        "resistance", 
        {"diffusion": {"kappa2": 1}}, 
        {"diffusion": {"kappa2": 10}}
    ], 
    score_pos="top"
)

ax.set_axis_off()
plt.show()
```
